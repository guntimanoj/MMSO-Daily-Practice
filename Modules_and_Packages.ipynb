```python
%%writefile student.py

def calculate_average(marks):
  if not marks:
    raise ValueError("Marks cannot be empty")

  return sum(marks) / len(marks)

def get_result(marks):
  average = calculate_average(marks)

  if average >= 50:
    return "Pass"
  else:
    return "Fail"
```

```python
%%writefile report.py

def display_report(name, marks, average, result):
  print("Student;", name)
  print("Marks:", marks)
  print("Average:", round(average, 2))
  print("Result:", result)
```

```python
%%writefile main.py

from student import calculate_average, get_result
from report import display_report

name = "Alex"
marks = [75, 80, 65]

average = calculate_average(marks)
result = get_result(marks)

display_report(name, marks, average, result)
```
