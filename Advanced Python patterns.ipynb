```python
marks = [78, 85, 32, 91, 45]

for index, mark in enumerate(marks):
  if mark < 50:
    print(f"Failed mark: {marks}, index: {index}")

print("Congratulations!")
```

```python
#Asscending order
print("Ascending order:")
numbers = [40, 10, 30, 20]
result = sorted(numbers)
print(result)
print(numbers)
#Decending order
print("Descending order:")
answer = sorted(numbers, reverse=True)
print(answer)
print(numbers)
```

```python
marks = [35, 42, 78, 91]
print(any(mark >= 90 for mark in marks))
print(any(mark < 50 for mark in marks))
```

```python
marks = [75, 82, 90]

print(all(mark >= 50 for mark in marks))
print(all(mark >= 90 for mark in marks))
```

```python
print(isinstance(25, int))
print(isinstance(2.3, float))
print(isinstance("Python", str))
print(isinstance(25, str))
```

```python
numbers = [10, 25, 40, 25, 40, 30]
unique_numbers = sorted(set(numbers), reverse=True)

if len(unique_numbers) >= 2:
  print(unique_numbers[1])
else:
  print("No second largest number")
```

```python
marks = [65, 82, 45, 91, 73]

passed = [mark for mark in marks if mark >= 50]

print("Passed:", passed)
print("Highest:", max(marks))
print("Lowest:", min(marks))
print("Average:", sum(marks) / len(marks))
print("Everyone Passed:", all(mark >= 50 for mark in marks))
print("Someone scored above 90:", any(mark > 90 for mark in marks))
```

```python
employees = [
    {"name": "Anu", "Salary": 2500},
    {"name": "Rahul", "Salary": 3000},
    {"name": "Manoj", "Salary": 9000},
    {"name": "Ravi", "Salary": 5000}
]
ranked = sorted(employees, key=lambda employee: employee["Salary"], reverse=True)
for rank, employee in enumerate(ranked, start=1):
  print(f"{rank}. {employee['name']} {employee['Salary']}")
```
