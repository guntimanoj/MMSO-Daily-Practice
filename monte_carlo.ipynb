```python
import numpy as np
rng = np.random.default_rng()

n = 1_000_000
pts = rng.random((n, 2))
inside = (pts ** 2).sum(axis=1) <= 1
print(f"Pi estimate: {4 * inside.mean():.4f} (true 3.1416)")

d1 = rng.integers(1, 7, 500_000)
d2 = rng.integers(1, 7, 500_000)
print(f"(sum = 7): {(d1 + d2 == 7).mean():.4f}")

trials = 100_000
birthdays = rng.integers(0, 365, size=(trials, 23))
sorted_b = np.sort(birthdays, axis=1)
shared = (np.diff(sorted_b, axis=1) == 0).any(axis=1)
print(f"P(shared birthday among 23): {shared.mean():.3f}")
```
