```python
import numpy as np

def program_1_student_marks():
  print("\n===PROGRAM 1:Student marks analyzer===")
  names = np.array(["Asha", "Ravi", "Meera", "Karan", "Sita"])
  subjects = ["Maths", "Physics", "CS"]
  # Updated marks matrix to have 5 rows corresponding to the 5 students
  marks = np.array([
      [85, 78, 92],
      [67, 56, 56],
      [56, 74, 93],
      [90, 88, 85],
      [45, 52, 48]
  ])
  avg_per_student = marks.mean(axis=1)
  avg_per_subject = marks.mean(axis=0)
  topper = names[avg_per_student.argmax()]

  grades = np.where(avg_per_student >= 85, "A", np.where(avg_per_student >= 70, "B", np.where(avg_per_student >= 50, "C", "F")))

  print("Average per student:")
  for n, a, g in zip(names, avg_per_student, grades):
    print(f" {n:<6} {a:6.2f} grade {g}")
  for s, a in zip(subjects, avg_per_subject):
    print(f"Subject average {s}: {a:.2f}")
  print("Topper:", topper)
  print("Students failing any subject:", names[(marks < 50).any(axis=1)])
  print("Above class average:", names[avg_per_student > avg_per_student.mean()])

# Run the program
program_1_student_marks()
```

```python
import numpy as np

def program_2_stock_analysis():
  print("\n=== PROGRAM 2: STOCK price analysis===")
  rng = np.random.default_rng(42)

  daily_ret = rng.normal(loc=0.0005, scale=0.01, size=252)
  prices = 100 * np.cumprod(1 + daily_ret)

  returns = np.diff(prices) / prices[:-1]

  window = 20
  ma = np.convolve(prices, np.ones(window) / window, mode="valid")

  vol = returns.std(ddof=1) * np.sqrt(252)
  sharpe = returns.mean() / returns.std(ddof=1) * np.sqrt(252)

  running_max = np.maximum.accumulate(prices)
  drawdown = (prices - running_max) / running_max
  max_dd = drawdown.min()

  print(f"Start price: {prices[0]:.2f}, end price: {prices[-1]:.2f}")
  print(f"Total return: {(prices[-1] / prices[0] - 1) * 100:.2f}%")
  print(f"Last 20-day MA: {ma[-1]:.2f}")
  print(f"Annualised volatility: {vol * 100:.2f}%")
  print(f"Sharpe ratio: {sharpe:.2f}")
  print(f"Max drawdown: {max_dd * 100:.2f}%")
  print(f"Best day: {returns.max() * 100:.2f}% worst day: {returns.min() * 100:.2f}%")
  print("Day price was above its 20-day MA:", int((prices[window - 1:] > ma).sum()))

# Run the program
program_2_stock_analysis()
```
