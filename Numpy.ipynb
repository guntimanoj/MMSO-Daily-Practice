```python
import numpy as np

def show(title, val):
  print(f"{title}:\n{val}\n")

def program_broadcasting():
  print("\n" + "=" * 60)
  print("=" * 60)

  a = np.array([1, 2, 3])
  show("a+10", a + 10)

  m = np.array([[1, 2, 3],
                [4, 5, 6]])
  row = np.array([10, 20, 30])
  col = np.array([[100], [200]])
  show("m + row", m + row)
  show("m + col", m + col)

  print("Shape compatibility: ")
  for s1, s2 in [((2, 3), (3,)), ((2, 3), (2, 1)), ((3, 1), (1, 4)), ((2, 3), (2,))]:
    try:
      print(f" {s1} with {s2} -> {np.broadcast_shapes(s1, s2)}")
    except:
      print(f" {s1} with {s2} -> ERROR: not compatible")

# Run the demonstration function
program_broadcasting()
print()

nums = np.arange(1, 6)
table = nums[:, np.newaxis] * nums[np.newaxis, :]
show("Multiplication table", table)

scores = np.array([[80, 70, 90],
                   [60, 65, 70],
                   [100, 90, 80],
                   [40, 50, 60]])
col_means = scores.mean(axis=0)
show("Column means", col_means)
show("Centred scores", scores - col_means)

row_sums = scores.sum(axis=1, keepdims=True)
show("Each subject as % of student total", np.round(scores / row_sums * 100, 1))

pts = np.array([[0, 0], [3, 4], [6, 8]])
diff = pts[:, np.newaxis, :] - pts[np.newaxis, :, :]
dist = np.sqrt((diff ** 2).sum(axis=2))
show("Pairwise distance matrix", dist)
```
