```python
add = lambda a, b: a + b
print(add(10, 20))
is_even = lambda n: n % 2 == 0
print(is_even(10))
positive = lambda n: n >= 0
print(positive(5))
get_name = lambda student: student["name"]
print(get_name({"name": "Manoj"}))
square = lambda n: n * n
print(square(5))
is_odd = lambda n: n % 2 == 1
print(is_odd(6))
```

```python
numbers = [1, 2, 3, 4, 5]
squares = map(lambda n: n * n , numbers)
print(list(squares))
```

```python
marks = [35, 72, 48, 90, 29, 65]
passed = list(filter(lambda mark: mark >= 40, marks))
print(passed)
```

```python
names = ["Manoj", "Raju", "Mallaiah"]
marks = [60, 95, 90]
result = list(zip(names, marks))
print(result)
```
