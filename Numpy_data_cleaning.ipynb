```python
import numpy as np

def data_cleaning():
  data = np.array([23, 25, np.nan, 22, 24, 250, 26, np.nan, 21, 27, -40], dtype=float)
  print("Raw data:", data)
  print("Missing Values:", np.isnan(data).sum())

  q1, q3 = np.nanpercentile(data, [25, 75])
  iqr = q3 - q1
  low, high = q1 - 1.5 * iqr, q3 + 1.5 * iqr
  outlier_mask = (data < low) | (data > high)
  print(f"IQR bounds: [{low:.1f}, {high:.1f}], outliers: {data[outlier_mask]}")

  cleaned = np.where(outlier_mask, np.nan, data)

  fill = np.nanmean(cleaned)
  cleaned = np.where(np.isnan(cleaned), fill, cleaned)
  print("After filling:", np.round(cleaned, 2))

  minmax = (cleaned - cleaned.min()) / (cleaned.max() - cleaned.min())
  zscore = (cleaned - cleaned.mean()) / cleaned.std()
  print("Min-max:", np.round(minmax, 2))
  print("Z-score mean - 0:", round(zscore.mean(), 10), ", std:", round(zscore.std(), 2))

data_cleaning()
```
