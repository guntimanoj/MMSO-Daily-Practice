```python
data = [10, "Python", 20.5, 30, "SQL", 40.2, 50]

integer_count = 0
string_count = 0
float_count = 0

for item in data:
  if type(item) == int:
    integer_count += 1
  elif type(item) == str:
    string_count += 1
  elif type(item) == float:
    float_count += 1

print("Integers:", integer_count)
print("Strings:", string_count)
print("Float:", float_count)
```

```python
def safe_int(s):
  try:
    return int(s)
  except (ValueError, TypeError):
    return None
print(safe_int("42"),safe_int("abc"))
```
