```python
import numpy as np

def linear_regression():
  rng = np.random.default_rng(0)

  x = rng.uniform(0, 10, 100)
  y = 3 + 2 * x + rng.normal(0, 1.5, 100)

  X = np.column_stack([np.ones_like(x), x])

  beta = np.linalg.solve(X.T @ X, X.T @ y)
  print(f"Normal equation -> intercept: {beta[0]:.3f}, slope: {beta[1]:.3f}")

  beta_ls, *_ = np.linalg.lstsq(X, y, rcond=None)
  print(f"np.linalg.lstsq -> intercept: {beta_ls[0]:.3f}, slope: {beta_ls[1]:.3f}")

  y_pred = X @ beta
  ss_res = ((y - y_pred) ** 2).sum()
  ss_tot = ((y - y.mean()) ** 2).sum()
  r2 = 1 - ss_res / ss_tot
  rmse = np.sqrt(((y - y_pred) ** 2).mean())
  print(f"R^2: {r2:.4f}, RMSE: {rmse:.3f}")
  print(f"Correlation x,y: {np.corrcoef(x, y)[0, 1]:.4f}")
  print(f"Prediction for x=7: {beta[0] + beta[1] * 7:.2f}")

linear_regression()
```
