```python
squares = [number ** 2 for number in range(1, 11)]
print(squares)
```

```python
even_numbers = [
    number for number in range(1, 11)
    if number % 2 == 0
]
print(even_numbers)
```

```python
marks = [35, 80, 45, 90, 60]
results = ["Pass" if mark >= 50 else "Fail" for mark in marks]
print(results)
```

```python
squares = {number: number ** 2 for number in range(1, 11)}
print(squares)
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 5, 6, 6]
unique_squares = {number ** 2 for number in numbers}
print(unique_squares)
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
flattened = [number for row in matrix for number in row]
print(flattened)
```

```python
words = ["Python", "SQL", "Pandas"]
lengths = {word: len(word) for word in words}
print(lengths)
```
