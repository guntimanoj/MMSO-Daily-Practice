```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2 * np.pi, 2 * np.pi, 200)

y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(10, 2), facecolor='white')
plt.plot(x, y1, color=(0.10, 0.4470, 0.7410), linewidth=2, label='Dataset 1 (sin)')
plt.plot(x, y2, color=(0.8500, 0.3250, 0.0990), linestyle='--', linewidth=2, label='Dataset 2 (cos)')

plt.title('My Graph', fontsize=10, fontweight='bold')
plt.xlabel('X-Axis Label (units)', fontsize=12)
plt.ylabel('Y-Axis Label (units)', fontsize=12)

plt.legend(loc='best')
plt.grid(True, linestyle=':', alpha=0.9)
plt.xlim(x.min(), x.max())
plt.show()
```
